```python
class Book:
    def __init__(self, Book_id, Book_Name):
        self.Book_id = Book_id
        self.Book_Name = Book_Name
     
    def displayBook(self):
        print("Book_id : ", self.Book_id, ", Book_Name: ", self.Book_Name)
 
# Initial Library Data (Note: IDs are integers here)
library = [
    Book(1, "Bharat"),
    Book(2, "The Hobbit"),
    Book(3, "Python Basics")
]

print("--- Displaying All Books ---")
for book in library:
    book.displayBook() 
     
def add_newBook_to_Library(library):
    print("\n---- Adding New Book ----")
    # Cast to int to match the initial library data IDs
    NewBook_id = int(input("Enter new Book ID: "))
    NewBook_name = input("Enter new Book Name: ")

    New_Book = Book(NewBook_id, NewBook_name)
    library.append(New_Book)
     
add_newBook_to_Library(library)       

print("\n--- Displaying Updated Books ---")
for book in library:
    book.displayBook() 

def delete_book(library):
    print("\n----- Delete Book -------")
    Delete_id = int(input("Enter Book ID to delete: ")) # Fixed missing parenthesis
    
    # Search for the book with the matching ID
    book_found = False
    for book in library:
        if book.Book_id == Delete_id:
            library.remove(book)
            print(f"Success: '{book.Book_Name}' has been deleted.")
            book_found = True
            break # Exit the loop once the book is removed
            
    if not book_found:
        print("Error: Book ID not found.")

# Test the delete function
delete_book(library)

print("\n--- Displaying Library After Deletion ---")
for book in library:
    book.displayBook()
    
```
