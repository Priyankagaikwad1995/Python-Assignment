```python
queue = []
while True:
    print("1. Add a member to the Queue")
    print("2. Remove member from the queue")
    print("3. Display the current members")
    print("4. Quit")
    ch=input("Enter your choice: ")
    if ch=="1":
        val = input("Enter the value to be added in the queue:")
        queue.append(val)
        
    elif ch=="2":
        if len(queue)<1:
            print("Queue is Empty!")
            continue
        queue.pop(0)
        
    elif ch=="3":
        if len(queue)<1:
            print("Queue is Empty!")
            continue
        print("Members of the queue are:")
        for i in queue:
            print(i, end=" - ")
        print()
    elif ch=="4":
        print("Thank you")
        break
    else:
        print("Invalid Opton, try again!")
        
```

```python
stack = []

while True:
  print("1. Add a member to the stack")
  print("2. Remove member from the stack")
  print("3. Display the current members")
  print("4. Quit")

  ch = input("Enter your choice: ")

  if ch == "1":
    val = input("Enter the value to be added in the stack: ")
    stack.append(val)

  elif ch == "2":
    if len(stack) < 1:
      print("Stack is Empty!")
      continue
    removed_val = stack.pop()  # Removes the last element (LIFO)
    print(f"Removed member: {removed_val}")

  elif ch == "3":
    if len(stack) < 1:
      print("Stack is Empty!")
      continue
    print("Members of the stack are:")
    for i in stack:
      print(i, end="-")
    print()

  elif ch == "4":
    print("Thank you")
    break

  else:
    print("Invalid Option, try again!")
```
